```python
import torch
import numpy as np
```

```python
data_total = torch.load('/pscratch/sd/z/zhantao/neurorient_repo/data/PR772/PR772_neurorient_downsampled_128x128.pt')
```

```python
print(data_total.keys())
```

```python
for k, v in data_total.items():
    if isinstance(v, torch.Tensor) or isinstance(v, np.ndarray):
        print(k, v.shape)
```

```python
seed = 42

total_num_images = data_total['intensities'].size(0)
```

```python
from sklearn.model_selection import train_test_split
```

```python
indices_set1, indices_set2 = train_test_split(np.arange(total_num_images), test_size=0.5, random_state=seed)
```

```python
data_set1 = {}
data_set2 = {}
```

```python
for k, v in data_total.items():
    if k != 'intensities':
        data_set1[k] = v
        data_set2[k] = v
    else:
        data_set1[k] = v[indices_set1]
        data_set2[k] = v[indices_set2]
```

```python
data_set1.keys()
```

```python
torch.save(
    data_set1,
    '/pscratch/sd/z/zhantao/neurorient_repo/data/PR772/split/PR772_neurorient_downsampled_128x128_set1.pt'
)

torch.save(
    data_set2,
    '/pscratch/sd/z/zhantao/neurorient_repo/data/PR772/split/PR772_neurorient_downsampled_128x128_set2.pt'
)
```
